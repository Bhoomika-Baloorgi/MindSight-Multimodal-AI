```python
# ============================================================
#   MINDSIGHT DEMO — Master Single-Cell Runner
# ============================================================
import os, json, pickle, warnings, subprocess, sys, re
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from PIL import Image
from datetime import datetime
from google.colab import drive, files
import IPython.display as ipd

warnings.filterwarnings('ignore')

# ── 1. Install Dependencies ────────────────────────────────────
print('📦 Setting up MindSight dependencies (takes ~25s)...')
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q',
    'transformers', 'groq', 'librosa', 'soundfile', 'Pillow',
    'scikit-learn', 'torch', 'torchvision', 'torchaudio',
    'accelerate', 'opencv-python'], check=False)

import cv2
import torch
import librosa
from transformers import pipeline

# ── 2. Mount Google Drive ─────────────────────────────────────
try:
    drive.mount('/content/drive', force_remount=False)
except Exception as e:
    print(f'Drive note: {e}')

# ── 3. Configuration & Constants ──────────────────────────────
DEVICE = 0 if torch.cuda.is_available() else -1
GROQ_KEY = 'YOUR_GROQ_API_KEY'

EMOTIONS = ['angry', 'disgust', 'fear', 'happy', 'neutral', 'sad', 'surprise']
EMO_TO_STATE = {
    'angry': 'negative', 'disgust': 'negative', 'fear': 'negative', 'sad': 'negative',
    'neutral': 'neutral',
    'happy': 'positive', 'surprise': 'positive',
}
EMO_EMOJI = {'angry': '😠', 'disgust': '🤢', 'fear': '😨', 'happy': '😊', 'neutral': '😐', 'sad': '😢', 'surprise': '😲'}
STATE_COL = {'positive': '#2ECC71', 'neutral': '#95A5A6', 'negative': '#E74C3C'}
WEIGHTS   = {'face': 0.40, 'audio': 0.30, 'text': 0.20, 'behaviour': 0.10}

print('\n' + '='*55)
print('  MINDSIGHT — Loading Calibrated AI Models...')
print('='*55)

# ── 4. Load Models ────────────────────────────────────────────

# A. Face Model: SOTA Vision Transformer (ViT) + OpenCV Haar Cascade Cropper
print('📷 Loading Face Emotion ViT Model...')
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
face_pipe = pipeline('image-classification', model='trpakov/vit-face-expression', device=DEVICE)
print('  ✓ Face ViT model loaded')

# B. Text Model: DistilRoBERTa Emotion Classifier
print('📝 Loading Text Emotion RoBERTa Model...')
text_pipe = pipeline('text-classification', model='j-hartmann/emotion-english-distilroberta-base', top_k=None, device=DEVICE)
print('  ✓ Text model loaded')

# C. Audio Model: Wav2Vec2 Speech Emotion Recognition
print('🎙️  Loading Speech Emotion Model...')
audio_pipe = pipeline('audio-classification', model='ehcalabres/wav2vec2-lg-xlsr-en-speech-emotion-recognition', device=DEVICE)
print('  ✓ Speech Wav2Vec2 model loaded')

print('\n✅ All models loaded and calibrated for demo!\n')

# ── 5. Lexicons & Calibrated Prediction Engines ───────────────

LEXICON = {
    'angry': ['suck', 'sucks', 'hate', 'furious', 'rage', 'angry', 'traffic', 'idiot', 'pissed', 'annoying', 'terrible', 'worst', 'stupid', 'mad', 'damn', 'hell', 'irritated'],
    'disgust': ['disgusting', 'gross', 'nasty', 'repulsive', 'sick', 'awful', 'filthy', 'yuck', 'creepy'],
    'fear': ['scared', 'afraid', 'terrified', 'anxious', 'panic', 'nervous', 'worry', 'worried', 'stress', 'stressed', 'dread', 'frightened'],
    'sad': ['sad', 'depressed', 'crying', 'cry', 'lonely', 'hopeless', 'heartbroken', 'miserable', 'grief', 'unhappy', 'tired', 'exhausted', 'overwhelmed'],
    'happy': ['happy', 'great', 'awesome', 'wonderful', 'joy', 'love', 'excited', 'good', 'fantastic', 'blessed', 'proud', 'glad', 'celebrate'],
    'surprise': ['wow', 'shocked', 'surprised', 'unbelievable', 'omg', 'unexpected', 'astonished', 'crazy'],
    'neutral': ['okay', 'fine', 'normal', 'routine', 'average', 'standard', 'regular']
}

def to_7vec(raw_dict):
    MAPPING = {
        'anger': 'angry', 'angry': 'angry',
        'disgust': 'disgust',
        'fear': 'fear', 'fearful': 'fear',
        'joy': 'happy', 'happy': 'happy', 'happiness': 'happy',
        'neutral': 'neutral', 'calm': 'neutral',
        'sadness': 'sad', 'sad': 'sad',
        'surprise': 'surprise', 'surprised': 'surprise'
    }
    vec = np.zeros(7, dtype=np.float32)
    for k, v in raw_dict.items():
        canon = MAPPING.get(k.lower().strip())
        if canon in EMOTIONS:
            vec[EMOTIONS.index(canon)] += float(v)
    s = vec.sum()
    return (vec / s).astype(np.float32) if s > 0 else (np.ones(7, dtype=np.float32) / 7.0)

def predict_face(img_path):
    img_bgr = cv2.imread(img_path)
    if img_bgr is None:
        img_pil = Image.open(img_path).convert('RGB')
        img_bgr = cv2.cvtColor(np.array(img_pil), cv2.COLOR_RGB2BGR)

    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=3, minSize=(35, 35))

    if len(faces) > 0:
        x, y, w, h = max(faces, key=lambda f: f[2] * f[3])
        mx, my = int(w * 0.15), int(h * 0.15)
        x1, y1 = max(0, x - mx), max(0, y - my)
        x2, y2 = min(img_bgr.shape[1], x + w + mx), min(img_bgr.shape[0], y + h + my)
        face_crop = img_bgr[y1:y2, x1:x2]
        face_pil = Image.fromarray(cv2.cvtColor(face_crop, cv2.COLOR_BGR2RGB))
    else:
        face_pil = Image.fromarray(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))

    res = face_pipe(face_pil)
    raw = {item['label']: item['score'] for item in res}
    vec = to_7vec(raw)
    vec = np.exp(np.log(vec + 1e-6) * 1.35)
    return (vec / vec.sum()).astype(np.float32)

def predict_text(text):
    res = text_pipe(text)[0]
    raw = {item['label']: item['score'] for item in res}
    vec = to_7vec(raw)

    words = re.findall(r'\w+', text.lower())
    boosts = np.zeros(7, dtype=np.float32)
    for emo_idx, emo_name in enumerate(EMOTIONS):
        matches = sum(1 for w in words if w in LEXICON.get(emo_name, []))
        if matches > 0:
            boosts[emo_idx] += (0.45 * matches)

    vec = vec + boosts
    return (vec / vec.sum()).astype(np.float32)

def predict_audio(audio_path):
    try:
        y, sr = librosa.load(audio_path, sr=16000, mono=True)
        y = (y - np.mean(y)) / (np.std(y) + 1e-7)
        res = audio_pipe(audio_path)
        raw = {item['label']: item['score'] for item in res}
        vec = to_7vec(raw)
        vec = np.exp(np.log(vec + 1e-6) * 1.25)
        return (vec / vec.sum()).astype(np.float32)
    except Exception as e:
        print(f'  Audio notice: {e}')
        return None

def predict_behaviour(r):
    """
    Holistic Behavioural & Lifestyle Model.
    Balances sleep quality & duration, productivity buffers, exercise, and screen fatigue.
    """
    sleep_q = float(r.get('sleep_quality_1_5', 3))
    sleep_h = float(r.get('sleep_hours', 7.0))
    screen_h = float(r.get('screen_time_hours', 6.0))
    exercise = float(r.get('exercise_minutes_per_week', 60))
    productivity = float(r.get('productivity_0_100', 60))

    # 1. Stress Burdens
    sleep_def_penalty = max(0.0, 7.0 - sleep_h) * 1.8
    sleep_qual_penalty = max(0.0, 3.0 - sleep_q) * 1.8
    screen_strain_penalty = min(3.0, max(0.0, screen_h - 8.0) * 0.4) # capped strain
    low_prod_penalty = max(0.0, 40.0 - productivity) * 0.05
    stress_burden = sleep_def_penalty + sleep_qual_penalty + screen_strain_penalty + low_prod_penalty

    # 2. Resilience & Wellness Buffers (Protective Factors)
    good_sleep_buffer = max(0.0, sleep_h - 7.0) * 0.8 + max(0.0, sleep_q - 3.0) * 1.2
    high_prod_buffer = max(0.0, productivity - 60.0) * 0.04
    exercise_buffer = min(2.0, exercise / 50.0)
    resilience = good_sleep_buffer + high_prod_buffer + exercise_buffer

    # 3. Net Stress Index
    net_stress = max(0.0, stress_burden - resilience)

    # 4. Accurate Emotional Distribution Mapping
    if net_stress >= 5.0:
        stress = 'High'
        # High Burnout / Exhaustion / Strain
        vec = np.array([0.25, 0.10, 0.25, 0.02, 0.08, 0.28, 0.02], dtype=np.float32)
    elif net_stress >= 2.0:
        stress = 'Medium'
        # Moderate fatigue / Mild pressure
        vec = np.array([0.10, 0.05, 0.12, 0.15, 0.35, 0.20, 0.03], dtype=np.float32)
    else:
        stress = 'Low'
        if productivity >= 70 and screen_h >= 10:
            # Focused / High Deep Work / Productive Flow
            vec = np.array([0.02, 0.01, 0.02, 0.38, 0.50, 0.05, 0.02], dtype=np.float32)
        else:
            # Balanced / Relaxed
            vec = np.array([0.02, 0.01, 0.02, 0.48, 0.40, 0.04, 0.03], dtype=np.float32)

    return (vec / vec.sum()).astype(np.float32), stress

def fuse(d):
    active  = {k: v for k, v in d.items() if v is not None}
    total_w = sum(WEIGHTS[k] for k in active)
    fused   = np.zeros(7, dtype=np.float32)
    used_w  = {}
    for k, p in active.items():
        nw        = WEIGHTS[k] / total_w
        fused    += nw * p
        used_w[k] = round(nw, 3)
    return fused / (fused.sum() + 1e-8), used_w

# ── 6. Dynamic, State-Specific Wellness Protocols ─────────────

EMOTION_PROTOCOLS = {
    'angry': {
        'desc': 'acute frustration, irritation, or heightened physiological tension.',
        'insight': 'Anger and frustration indicate that a personal expectation or boundary is experiencing friction. Discharging the somatic adrenaline safely restores analytical clarity.',
        'actions': [
            '💨 Execute the Physiological Sigh (2 quick deep inhales through nose, followed by 1 long slow exhale through mouth) 5 times to reset heart rate variability.',
            '🚶 Step away from screens for a 10-minute physical walk to clear adrenaline.',
            '📝 Do a 2-minute "unfiltered brain dump" on paper, then set it aside to free your working memory.'
        ]
    },
    'disgust': {
        'desc': 'strong sensory aversion, boundary violation, or environmental discomfort.',
        'insight': 'Disgust is your brain\'s boundary alarm signalling that your immediate physical, social, or digital environment feels misaligned with your standards.',
        'actions': [
            '🌿 Physically change your immediate environment (move to a fresh room or step outside into fresh air).',
            '💧 Drink a large glass of cold water and wash your hands or face for a sensory reset.',
            '🛡️ Identify the exact situational trigger and enforce a clean boundary around it today.'
        ]
    },
    'fear': {
        'desc': 'anticipatory anxiety, nervousness, or cognitive overwhelm regarding demands.',
        'insight': 'Anxiety occurs when the nervous system overestimates future threats while underestimating current capability and resilience.',
        'actions': [
            '👁️ Practice the 5-4-3-2-1 Grounding Method: Name 5 things you see, 4 you feel, 3 you hear, 2 you smell, and 1 positive truth about yourself.',
            '📦 Break whatever task feels heavy into a single 5-minute micro-step and ignore everything else for now.',
            '🫁 Use Box Breathing (Inhale 4s, Hold 4s, Exhale 4s, Hold 4s) for 3 consecutive cycles.'
        ]
    },
    'sad': {
        'desc': 'low emotional energy, grief, loneliness, or cognitive exhaustion.',
        'insight': 'Sadness is your nervous system\'s signal to slow down, conserve resources, and seek comfort rather than pushing through high demands.',
        'actions': [
            '☕ Wrap yourself in physical warmth (a warm tea, blanket, or warm shower) to support psychological safety.',
            '🤝 Send a short text to one trusted friend or family member just saying hello.',
            '☀️ Get 10 minutes of direct outdoor natural sunlight to support dopamine and serotonin synthesis.'
        ]
    },
    'happy': {
        'desc': 'high positive valence, emotional openness, and vitality.',
        'insight': 'Positive emotional states expand cognitive flexibility, problem-solving capability, and interpersonal connection.',
        'actions': [
            '⚡ Channel this energetic window into your most challenging analytical or creative project right now.',
            '💌 Send a quick message of appreciation or gratitude to someone who has supported you recently.',
            '🧠 Savor the moment: pause for 30 seconds to intentionally register how good this feeling feels in your body.'
        ]
    },
    'surprise': {
        'desc': 'cognitive disorientation, unexpected disruption, or sudden stimulation.',
        'insight': 'Surprise pauses standard neural routines. How you frame this unexpected event (threat vs opportunity) dictates your recovery speed.',
        'actions': [
            '🛑 Pause before making impulsive decisions or replying immediately to unexpected messages.',
            '📋 Write down the objective facts of what just occurred separately from your emotional interpretation.',
            '🧘 Take 3 slow belly breaths to allow your prefrontal cortex to regain full analytical control.'
        ]
    },
    'neutral': {
        'desc': 'calm, baseline emotional equilibrium, high focus, and steady cognitive control.',
        'insight': 'A neutral baseline is an ideal cognitive state for deep, focused execution and strategic planning without emotional distortion.',
        'actions': [
            '🎯 Establish a 45-minute Pomodoro deep-work block to make steady progress on high-priority goals.',
            '💧 Drink 500ml of water and do 3 full-body shoulder and neck stretches to maintain posture.',
            '📵 Keep unnecessary notifications muted to preserve this calm mental clarity.'
        ]
    }
}

def generate_ai_report(emotion, state, conf, fused, modalities, stress, text_in, used_w, r_dict=None):
    proto = EMOTION_PROTOCOLS.get(emotion, EMOTION_PROTOCOLS['neutral'])

    lines = []
    for k, p in modalities.items():
        if p is not None:
            extra = f" ['{text_in[:35]}']" if k == 'text' and text_in else \
                    f" (stress={stress})"   if k == 'behaviour' else ""
            lines.append(f"- {k.upper():<10}: {EMOTIONS[np.argmax(p)].upper()} ({p.max()*100:.1f}%){extra}")
    top3 = sorted(zip(EMOTIONS, fused), key=lambda x: -x[1])[:3]
    mod = '\n'.join(lines)

    actions = list(proto['actions'])
    if r_dict:
        if r_dict.get('screen_time_hours', 0) >= 10.0 and r_dict.get('sleep_hours', 7) >= 7.0:
            actions[1] = f"👀 Intensive Screen Day ({r_dict['screen_time_hours']}h): Practice the 20-20-20 rule (every 20 mins, look at an object 20 feet away for 20 seconds) to prevent eye fatigue."
        elif r_dict.get('sleep_hours', 7) <= 5.5:
            actions[0] = f"😴 Sleep Deficit ({r_dict['sleep_hours']}h last night): Practice a 15-minute Non-Sleep Deep Rest (NSDR) and avoid caffeine after 2 PM."

    try:
        from groq import Groq
        client = Groq(api_key=GROQ_KEY)
        prompt = f"""You are MindSight, an empathetic, non-clinical AI mental wellness analyst.

ANALYSIS RESULT:
- Primary emotion: {emotion.upper()} ({conf*100:.1f}% confidence)
- Emotional state: {state.upper()}
- Top 3 emotions : {', '.join(f'{e}({p*100:.1f}%)' for e,p in top3)}
- Modalities Used: {', '.join(f'{k}={v}' for k,v in used_w.items())}
- Modality Signals:
{mod}

Context Insight: {proto['insight']}
Specific Actions:
1. {actions[0]}
2. {actions[1]}
3. {actions[2]}

Write a concise, personalized, warm report in exactly 4 sections:
1. WHAT WE DETECTED
2. WHAT YOUR SIGNALS REVEAL
3. MENTAL HEALTH INSIGHT
4. 3 IMMEDIATE ACTIONS FOR YOU

Keep it under 260 words. Speak warmly as 'you'. Never use generic clichés or clinical diagnoses."""

        resp = client.chat.completions.create(
            model='llama3-8b-8192',
            messages=[{'role': 'user', 'content': prompt}],
            max_tokens=450
        )
        return resp.choices[0].message.content
    except Exception:
        return f"""1. WHAT WE DETECTED
Our multimodal analysis detected {emotion.upper()} ({conf*100:.1f}% confidence), indicating a {state.upper()} emotional baseline.

2. WHAT YOUR SIGNALS REVEAL
{mod}

3. MENTAL HEALTH INSIGHT
{proto['insight']}

4. 3 IMMEDIATE ACTIONS FOR YOU
1. {actions[0]}
2. {actions[1]}
3. {actions[2]}"""

def show_results(fused, modalities, emotion, state, conf):
    panels = {k: v for k, v in modalities.items() if v is not None}
    panels['FUSED ★'] = fused

    n_panels = len(panels)
    col = STATE_COL.get(state, '#3498DB')
    emoji = EMO_EMOJI.get(emotion, '')

    fig = plt.figure(figsize=(4.5 * n_panels, 6.5), dpi=150)
    gs  = gridspec.GridSpec(1, n_panels, wspace=0.35)

    fig.suptitle(
        f'MindSight Result  {emoji}  {emotion.upper()}  |  {state.upper()}  ({conf*100:.1f}% confidence)',
        fontsize=14, fontweight='bold', color=col, y=1.03
    )

    for idx, (name, proba) in enumerate(panels.items()):
        ax = fig.add_subplot(gs[idx])
        top_idx = int(np.argmax(proba))
        is_fused = 'FUSED' in name

        colors = [
            col       if (i == top_idx and is_fused) else
            '#1565C0' if i == top_idx else
            '#BDC3C7' if is_fused   else '#90CAF9'
            for i in range(7)
        ]

        bars = ax.barh(EMOTIONS, proba * 100, color=colors, edgecolor='#333', linewidth=0.5)
        ax.set_xlim([0, 115])
        ax.set_title(
            f"{name.upper()}\n{EMO_EMOJI.get(EMOTIONS[top_idx], '')} {EMOTIONS[top_idx].upper()} ({proba.max()*100:.1f}%)",
            fontsize=10.5,
            fontweight='bold' if is_fused else 'normal',
            color=col if is_fused else '#2C3E50'
        )
        ax.grid(axis='x', linestyle='--', alpha=0.3)
        ax.spines[['top', 'right']].set_visible(False)
        ax.tick_params(axis='y', labelsize=8.5)
        for bar, p in zip(bars, proba):
            ax.text(p * 100 + 2, bar.get_y() + bar.get_height() / 2, f"{p*100:.1f}%", va='center', fontsize=8, fontweight='bold')

    plt.tight_layout()
    plt.savefig('/content/mindsight_result.png', dpi=150, bbox_inches='tight')
    plt.show()

# ════════════════════════════════════════════════════════════
#   CONTINUOUS INTERACTIVE DEMO LOOP
# ════════════════════════════════════════════════════════════
print('╔══════════════════════════════════════════════════════╗')
print('║   🧠  MINDSIGHT — CALIBRATED MULTIMODAL DEMO         ║')
print('╠══════════════════════════════════════════════════════╣')
print('║  • Runs continuously round after round               ║')
print('║  • Press Enter to skip any input                     ║')
print('║  • Type "exit" at any prompt to stop session         ║')
print('╚══════════════════════════════════════════════════════╝')

session_count = 0

while True:
    session_count += 1
    print(f'\n🔵 ══════════════ ROUND #{session_count} ══════════════')

    face_p = text_p = beh_p = audio_p = None
    stress = text_in = face_path = None
    r_lifestyle = {}

    # ── INPUT 1: FACE ─────────────────────────────────────────
    print('\n📷 [1/4] FACE IMAGE (Upload photo / or press Enter to skip)')
    ans_face = input('    Upload face image? (y/n): ').strip().lower()
    if ans_face in ['exit', 'q', 'quit']:
        print('\n👋 Exiting MindSight Demo. Good luck tomorrow! 🧠')
        break
    if ans_face == 'y':
        print('    Opening file picker...')
        up = files.upload()
        if up:
            face_path = list(up.keys())[0]
            img = Image.open(face_path).convert('RGB')
            fig, ax = plt.subplots(1, 1, figsize=(3, 3))
            ax.imshow(img); ax.axis('off'); ax.set_title('Uploaded Image', fontsize=9)
            plt.tight_layout(); plt.show()

            face_p = predict_face(face_path)
            top = EMOTIONS[np.argmax(face_p)]
            print(f"    ✅ Face Detected: {EMO_EMOJI.get(top, '')} {top.upper()} ({face_p.max()*100:.1f}%)")

    # ── INPUT 2: AUDIO ────────────────────────────────────────
    print('\n🎙️  [2/4] VOICE RECORDING (Upload .ogg/.wav/.mp3 / or press Enter to skip)')
    ans_audio = input('    Upload audio file? (y/n): ').strip().lower()
    if ans_audio in ['exit', 'q', 'quit']:
        print('\n👋 Exiting MindSight Demo. Good luck tomorrow! 🧠')
        break
    if ans_audio == 'y':
        print('    Opening file picker...')
        up = files.upload()
        if up:
            af = list(up.keys())[0]
            try:
                w, sr0 = librosa.load(af, sr=None)
                ipd.display(ipd.Audio(w, rate=sr0))
            except Exception: pass

            audio_p = predict_audio(af)
            if audio_p is not None:
                top = EMOTIONS[np.argmax(audio_p)]
                print(f"    ✅ Audio Detected: {EMO_EMOJI.get(top, '')} {top.upper()} ({audio_p.max()*100:.1f}%)")

    # ── INPUT 3: TEXT ─────────────────────────────────────────
    print('\n📝 [3/4] TEXT JOURNAL (Type thoughts e.g. "traffic sucks in banglore" / or Enter to skip)')
    text_in = input('    ➤ ').strip()
    if text_in.lower() in ['exit', 'q', 'quit']:
        print('\n👋 Exiting MindSight Demo. Good luck tomorrow! 🧠')
        break
    if text_in:
        text_p = predict_text(text_in)
        top = EMOTIONS[np.argmax(text_p)]
        print(f"    ✅ Text Detected: {EMO_EMOJI.get(top, '')} {top.upper()} ({text_p.max()*100:.1f}%)")

    # ── INPUT 4: BEHAVIOUR ────────────────────────────────────
    print('\n📊 [4/4] LIFESTYLE TELEMETRY (Answer questions / or press Enter to skip)')
    ans_beh = input('    Fill lifestyle questionnaire? (y/n): ').strip().lower()
    if ans_beh in ['exit', 'q', 'quit']:
        print('\n👋 Exiting MindSight Demo. Good luck tomorrow! 🧠')
        break
    if ans_beh == 'y':
        def ask_val(prompt, default, cast=int):
            raw = input(f'      {prompt} [{default}]: ').strip()
            if raw.lower() in ['exit', 'q', 'quit']: return 'exit'
            try: return cast(raw) if raw else default
            except: return default

        sleep_q = ask_val('Sleep Quality (1=poor 5=great)', 3, int)
        sleep_h = ask_val('Sleep Duration (hrs last night)', 7.0, float)
        screen_h = ask_val('Total Screen Time (hrs today)', 6.0, float)
        exercise = ask_val('Exercise this week (minutes)', 60, int)
        productivity = ask_val('Productivity today (0-100)', 60, int)

        if 'exit' in [sleep_q, sleep_h, screen_h, exercise, productivity]:
            print('\n👋 Exiting MindSight Demo. Good luck tomorrow! 🧠')
            break

        r_lifestyle = {'sleep_quality_1_5': sleep_q, 'sleep_hours': sleep_h, 'screen_time_hours': screen_h,
                       'exercise_minutes_per_week': exercise, 'productivity_0_100': productivity}
        beh_p, stress = predict_behaviour(r_lifestyle)
        top = EMOTIONS[np.argmax(beh_p)]
        print(f"    ✅ Behavioural Status: {stress.upper()} STRESS → {EMO_EMOJI.get(top, '')} {top.upper()}")

    # ── VALIDATION & FUSION ───────────────────────────────────
    active = [x for x in [face_p, audio_p, text_p, beh_p] if x is not None]
    if not active:
        print('\n⚠️ No inputs provided in this round. Please provide at least one input.')
    else:
        modalities = {'face': face_p, 'audio': audio_p, 'text': text_p, 'behaviour': beh_p}
        fused, used_w = fuse(modalities)
        emotion = EMOTIONS[np.argmax(fused)]
        state   = EMO_TO_STATE[emotion]
        conf    = float(fused.max())
        emoji   = EMO_EMOJI.get(emotion, '')

        print('\n' + '═'*55)
        print(f'  {emoji}  RESULT: {emotion.upper()} ({conf*100:.1f}% confidence)')
        print(f'  VALENCE STATE: {state.upper()}  |  WEIGHTS: {used_w}')
        print('═'*55)

        show_results(fused, modalities, emotion, state, conf)

        print('\nGenerating personalized mental wellness report...')
        report = generate_ai_report(emotion, state, conf, fused, modalities, stress, text_in, used_w, r_lifestyle)

        print('\n' + '═'*55)
        print('  💬 EMPATHETIC WELLNESS REPORT')
        print('═'*55)
        print(report)
        print('═'*55)

    # ── PROMPT FOR NEXT ROUND ─────────────────────────────────
    print()
    next_step = input('🔁 Do you want to run another analysis? (Press Enter to continue / type "exit" to stop): ').strip().lower()
    if next_step in ['exit', 'q', 'quit', 'n', 'no']:
        print('\n👋 Thank you for using MindSight! ')
        break
```
