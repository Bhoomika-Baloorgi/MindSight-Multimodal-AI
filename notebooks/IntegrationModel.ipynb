```python
# CELL — READ NOTEBOOKS FROM DRIVE AND EXTRACT KEY INFO
import json, os
from google.colab import drive

drive.mount("/content/drive", force_remount=False)

# ── Paths to your notebooks ───────────────────────────────────
# Update these if paths are different
NOTEBOOKS = {
    "audio"    : "/content/drive/MyDrive/Colab Notebooks/Audio_Emotion_WavLM_Final.ipynb",
    "text"     : "/content/drive/MyDrive/Colab Notebooks/BERT_Emotion_BestAccuracy.ipynb",
    "behaviour": "/content/drive/MyDrive/Colab Notebooks/Behaviour_v2.ipynb",
}

def extract_code_from_notebook(path):
    """Extract all code cells from a notebook as one string."""
    if not os.path.exists(path):
        return f"FILE NOT FOUND: {path}"
    with open(path, "r", encoding="utf-8") as f:
        nb = json.load(f)
    code_cells = []
    for i, cell in enumerate(nb.get("cells", [])):
        if cell.get("cell_type") == "code":
            src = "".join(cell.get("source", []))
            if src.strip():
                code_cells.append(f"# ── CELL {i+1} ──\n{src}")
    return "\n\n".join(code_cells)

for name, path in NOTEBOOKS.items():
    print("\n" + "="*70)
    print(f"  NOTEBOOK: {name.upper()}")
    print(f"  Path    : {path}")
    print("="*70)
    code = extract_code_from_notebook(path)
    # Print first 3000 chars — enough to see imports, model def, inference
    print(code[:3000])
    print(f"\n... [total chars: {len(code)}]")
    print("="*70)
```

```python
# CELL — READ KEY NOTEBOOKS FROM CORRECT PATHS
import json, os
from google.colab import drive

drive.mount("/content/drive", force_remount=False)

NOTEBOOKS_DIR = "/content/drive/MyDrive/Colab Notebooks"

# Read these specific notebooks
TARGETS = [
    "IntegrationModel.ipynb",        # existing integration attempt
    "MindSight_ULTIMATE.ipynb",       # likely latest full pipeline
    "MindSight_v3_FINAL.ipynb",       # full pipeline
    "BERT_final.ipynb",               # text model
    "VoiceEmotion_Standalone.ipynb",  # audio model
]

def extract_code(path, max_chars=5000):
    if not os.path.exists(path):
        return "NOT FOUND"
    with open(path, "r", encoding="utf-8") as f:
        nb = json.load(f)
    cells = []
    for i, cell in enumerate(nb.get("cells", [])):
        if cell.get("cell_type") == "code":
            src = "".join(cell.get("source", []))
            if src.strip():
                cells.append(f"# CELL {i+1}\n{src}")
    full = "\n\n".join(cells)
    return full[:max_chars]

for nb_name in TARGETS:
    path = os.path.join(NOTEBOOKS_DIR, nb_name)
    print("\n" + "="*65)
    print(f"NOTEBOOK: {nb_name}")
    print("="*65)
    code = extract_code(path)
    print(code)

# Also read the model metadata files
print("\n" + "="*65)
print("MODEL METADATA FROM MindSight_Models/")
print("="*65)
MODELS_DIR = "/content/drive/MyDrive/MindSight_Models"
for fname in ["text_labels.json", "voice_labels.json",
              "facial_labels.json", "behavioral_labels.json",
              "behavioral_columns.json"]:
    fpath = os.path.join(MODELS_DIR, fname)
    if os.path.exists(fpath):
        with open(fpath) as f:
            print(f"\n{fname}:")
            print(json.dumps(json.load(f), indent=2))
```
